# XiDepth v2.0 — Automated Cloud-to-Cloud KITTI Raw Downloader

This utility notebook automates downloading KITTI Raw sequence archives directly from the official public Amazon S3 repository (`https://s3.eu-central-1.amazonaws.com/avg-kitti/raw_data/`) into Kaggle or Colab storage at 100+ MB/s. Zero local bandwidth needed.

```python
import os, urllib.request, zipfile

BASE_URL = "https://s3.eu-central-1.amazonaws.com/avg-kitti/raw_data/"
TARGET_DIR = "/kaggle/working/kitti_raw"
os.makedirs(TARGET_DIR, exist_ok=True)

# Key sample continuous drive for testing or partial training
sample_drives = [
    "2011_09_26_drive_0001_sync.zip",
    "2011_09_26_calib.zip",
]

for archive in sample_drives:
    url = BASE_URL + archive
    dest = os.path.join(TARGET_DIR, archive)
    print(f"Downloading {archive}...")
    urllib.request.urlretrieve(url, dest)
    print(f"Extracting {archive}...")
    with zipfile.ZipFile(dest, "r") as zip_ref:
        zip_ref.extractall(TARGET_DIR)
    os.remove(dest)
print("Download and extraction complete!")
```
